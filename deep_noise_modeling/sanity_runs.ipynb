```python
from vae_flax import VAEEncoder
from jax import random
import jax
```

```python
# Example usage:
input_dim = 100
latent_dim = input_dim
batch = 1

model = VAEEncoder(input_dim=input_dim, latent_dim=latent_dim, hidden_dims=(128,64, 64, 64))
rng = random.PRNGKey(0)
x_dummy = jax.random.normal(rng, (batch, input_dim))
```

```python
x_dummy.shape
```

```python
# init params
variables = model.init(rng, x_dummy)     # {'params': {...}}
params = variables['params']
```

```python
# apply
mu, std, logvar = model.apply({'params': params}, x_dummy)
```

```python
mu.shape
```

```python
mu.shape == (batch, latent_dim)
# std.shape == (batch, latent_dim)
```

```python
eps = jax.random.normal(rng, mu.shape)
z = mu + std * eps
```
